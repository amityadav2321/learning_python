```python
class BankAccount:
    bank_name = "Rubicon National Bank"
    def __init__(self,name,balance=0):
        print(f"[constructer called] Opening account for {name}")
        self.name = name
        self.balance = balance

    def deposit(self,amount):
        self.balance +=amount
        print(f"Deposited {amount}. New Balance {self.balance}")

    def withdraw(self,amount):
        if self.balance > amount:
            self.balance -= amount
            print(f"Withdraw {amount}. Remaining Balance {self.balance}")
        else:
            print("Insufficent Balance")



account = BankAccount("Amit",1000)
print(account.name)
print(account.balance)
account.deposit(500)
account.withdraw(2000)
account.withdraw(200)


```

```python
class Student:
    def __init__(self, name, course):
        print(f"[constructor called] Creating student: {name}")
        self.name = name
        self.course = course

    def introduce(self):
        return f"Hi, I am {self.name}, learning {self.course}"

    def add_marks(self, marks):
        self.marks = marks

    def get_average(self):
        return sum(self.marks) / len(self.marks)


s1 = Student("Amit", "Python")
s2 = Student("Abhi", "Web")

print(s1.introduce())
print(s2.introduce())

s1.add_marks([23, 43, 45, 56])
print(s1.marks)
print(s1.get_average())
```

```python
class Vechicle:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model

    def show_details(self):
        print(f"The brand name is {self.brand} and model is {self.model}")


class Toyota(Vechicle):
    def __init__(self,brand,model,price):
        super().__init__(brand,model)
        self.price=price

    def about(self):
        print(f"The brand name is {self.brand} and model name is {self.model} and its price is {self.price}")

class Tata(Vechicle):
    def __init__(self, brand, model,safety):
        super().__init__(brand, model)
        self.safety = safety

    def about(self):
        print(f"The brand name is {self.brand} and model name is {self.model} and its safety rating is {self.safety}")


t1 = Toyota("Toyota","Fortuner",50000000)
t1.show_details()
t1.about()

t2 = Tata("Tata","Safari","5⭐")
t2.show_details()
t2.about()
        
```

```python
class Vechicle:
    def __init__(self,brand,speed):
        self.brand=brand
        self.speed=speed

    def show_details(self):
        print(f"The brand name is {self.brand} and speed is {self.speed}.")


class Car(Vechicle):
    def __init__(self,brand,speed,safety):
        super().__init__(brand,speed)
        
        self.safety = safety

    def about(self):
        print(f"The brand name is {self.brand} and speed is {self.speed} and its safety rating is {self.safety}.")

class Bike(Vechicle):
    def __init__(self, brand, speed,price):
        super().__init__(brand, speed)
        self.price=price
        

    def about(self):
        print(f"The brand name is {self.brand} and speed is {self.speed} and its price is {self.price}.")


t1 = Car("Toyota","150km/hr","5⭐")
t1.show_details()
t1.about()

t2 = Bike("Honda","120km/hr","2 Lakh")
t2.show_details()
t2.about()
        
```

```python
class Shape:
    def area(self):
        return 0;

class Square(Shape):
    def __init__(self,side):
        self.side = side

    def area(self):
        return self.side**2

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return round(3.142*self.radius**2,2)

shapes = [Square(4),Circle(3)]
for shape in shapes:
    print(f"{shape.__class__.__name__} area: {shape.area()}")
```

```python
import sqlite3
```

```python
conn =  sqlite3.connect("training db")
cursor = conn.cursor()
print("Connected to database successfully")
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        course Text,
        marks INTEGER
    )
""")
conn.commit()
print("Table student created or already existed")
cursor.execute("DELETE FROM students")
conn.commit()
```

```python
cursor.execute(
    "INSERT INTO students (name, course, marks) VALUES (?, ?, ?)",
    ("Amit","Python",88),
)
cursor.execute(
    "INSERT INTO students (name, course, marks) VALUES (?, ?, ?)",
    ("Abhi","Python + AI",80),
)
conn.commit()
print("Values are added")
```

```python
cursor.execute("SELECT * FROM students WHERE name = ?", ("Amit",))

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
more_students = [
    ("Priya", "Web Dev", 95),
    ("Karan", "Python", 60),
]
cursor.executemany(
    "INSERT INTO students (name, course, marks) VALUES (?, ?, ?)",
    more_students,
)
conn.commit()  # nothing is permanently saved until you commit()!
print("Records inserted!")
```

```python
cursor.execute("SELECT * FROM students")   # * means "all columns"
print("--- All Students ---")
for row in cursor.fetchall():              # fetchall() -> list of ALL matching rows
    print(row)
```

```python
cursor.execute("SELECT name, marks FROM students WHERE marks > 70")
print("--- Students with marks > 70 ---")
for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("SELECT * FROM students ORDER BY marks DESC")
print("--- Students sorted by marks (highest first) ---")
for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute(
    "UPDATE students SET marks = ? WHERE name = ?",
    (95, "Aditya"),
)
conn.commit()
print("Aditya's marks updated!")

# fetchone() -> grabs just the single next matching row (instead of a list)
cursor.execute("SELECT * FROM students WHERE name = 'Aditya'")
print(cursor.fetchone())
```

```python
cursor.execute("DELETE FROM students WHERE name = ?", ("Karan",))
conn.commit()
print("Karan's record deleted!")

cursor.execute("SELECT * FROM students")
print("--- Final Students Table ---")
for row in cursor.fetchall():
    print(row)
```

```python
conn.close()
print("Connection closed.")
```
